```python
# Import statements
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
from matplotlib import colormaps
from mpl_toolkits.mplot3d import Axes3D
import matplotlib.dates as mdates
from datetime import datetime, timedelta
from scipy.optimize import linear_sum_assignment
import ruptures as rpt
```

```python
PELT_MODEL = "l1"
PELT_MIN_SIZE = 10
PELT_PEN = 20
PELT_JUMP = 5

def twosigma_ang(ts):
    ts = pd.Series(ts, dtype=float).copy()
    if len(ts):
        ts.iloc[0] = np.nan
    mean = ts.expanding(min_periods=1).mean()
    std = ts.expanding(min_periods=1).std(ddof=0)
    flags = np.flatnonzero((np.abs(ts - mean) > 2 * std).to_numpy())
    return flags[flags != 1]

def normalised_2sigma_ang(ts):
    ts = pd.Series(ts, dtype=float)
    mean = ts.rolling(window=50).mean()
    std = ts.rolling(window=50).std(ddof=0).replace(0, np.nan)
    return twosigma_ang((ts - mean) / std)

def angle_displacement(ts):
    ts = np.asarray(ts, dtype=float)
    angle = np.full(len(ts), np.nan)
    for i in range(1, len(ts)):
        denominator = np.linalg.norm(ts[i - 1]) * np.linalg.norm(ts[i])
        if np.isfinite(denominator) and denominator > 0:
            cosine = np.dot(ts[i - 1], ts[i]) / denominator
            angle[i] = np.arccos(np.clip(np.abs(cosine), 0, 1))
    return angle

def eigen_decomp(cov):
    # eigh orders eigenvalues: always take the instantaneous largest mode.
    dates = cov.index.get_level_values(0).unique()
    size = cov.shape[1]
    cov_total = np.full(len(dates), np.nan)
    cov_var = np.full(len(dates), np.nan)
    cov_evec = np.full((len(dates), size), np.nan)
    cov_evec2 = np.full((len(dates), size), np.nan)
    for i, date in enumerate(dates):
        matrix = cov.loc[date].to_numpy(dtype=float)
        if not np.isfinite(matrix).all():
            continue
        values, vectors = np.linalg.eigh(matrix)
        cov_total[i] = values[-1]
        norm = np.linalg.norm(values)
        if norm > 0:
            cov_var[i] = values[-1] / norm
            cov_evec[i] = vectors[:, -1]
            if size > 1:
                cov_evec2[i] = vectors[:, -2]
    return cov_total, cov_var, cov_evec, cov_evec2

def vec_align(vectors):
    # Resolve only the arbitrary +/- sign; never swap eigenvector columns.
    aligned = np.array(vectors, dtype=float, copy=True)
    for i in range(1, len(aligned)):
        if np.isfinite(aligned[i - 1:i + 1]).all():
            if np.dot(aligned[i - 1], aligned[i]) < 0:
                aligned[i] *= -1
    return aligned

def valid_interval(signal):
    x = np.asarray(signal, dtype=float)
    if x.ndim == 1:
        x = x[:, None]
    valid = np.flatnonzero(np.isfinite(x).all(axis=1))
    if not len(valid):
        return x, 0, 0
    start, stop = valid[0], valid[-1] + 1
    return x, start, stop

def find_bkps(signal, min_size=PELT_MIN_SIZE, pen=PELT_PEN):
    """Exclusive segment endpoints in the original row coordinates, including stop."""
    x, start, stop = valid_interval(signal)
    if stop == start:
        return np.array([], dtype=int)
    centred = x[start:stop] - np.median(x[start:stop], axis=0)
    # A single scalar preserves relative component weights. The no-change
    # L1 cost is now one per observation, for both counts and vectors.
    scale = np.abs(centred).sum(axis=1).mean()
    if scale == 0 or stop - start < 2 * min_size:
        return np.array([stop], dtype=int)
    algo = rpt.Pelt(model=PELT_MODEL, min_size=min_size, jump=PELT_JUMP)
    endpoints = algo.fit(centred / scale).predict(pen=pen)
    return start + np.asarray(endpoints, dtype=int)

def shade_segments(ax, dates, vectors, bkps):
    # Interior endpoints mark the first observation in the next segment.
    dates = pd.DatetimeIndex(dates)
    _, start, stop = valid_interval(vectors)
    if stop == start:
        return
    edges = np.r_[start, bkps]
    colours = ("#4286f4", "#f44174")
    for i, (left, right) in enumerate(zip(edges[:-1], edges[1:])):
        end = dates[right] if right < stop else dates[stop - 1]
        ax.axvspan(dates[left], end, facecolor=colours[i % 2], alpha=0.2)

def get_signals_and_bkps(evectors_100, evectors_60):
    ang_100 = angle_displacement(evectors_100)
    ang_60 = angle_displacement(evectors_60)
    return (twosigma_ang(ang_100), normalised_2sigma_ang(ang_100),
            find_bkps(vec_align(evectors_100)),
            twosigma_ang(ang_60), normalised_2sigma_ang(ang_60),
            find_bkps(vec_align(evectors_60)))

def eigenvector_rotation_rate(vectors, dates):
    """Theory's consecutive leading-vector angle, divided by elapsed days.
    Uses the signed eigenvectors before componentwise absolute-value plotting.
    """
    dates = pd.DatetimeIndex(pd.to_datetime(dates))
    rate = angle_displacement(vectors)
    if len(dates) > 1:
        elapsed_days = np.asarray((dates[1:] - dates[:-1]) / pd.Timedelta(days=1))
        rate[1:] /= elapsed_days
    return rate

def get_signals_and_bkps_30(evectors_30):
    ang_30_raw = angle_displacement(evectors_30)
    return (twosigma_ang(ang_30_raw), normalised_2sigma_ang(ang_30_raw),
            find_bkps(vec_align(evectors_30)))
```
