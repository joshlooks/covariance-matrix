```python
def is_constant(series):
    return np.all(series == series[0])

def is_low_variance(series, threshold=1e-5):
    return np.var(series) < threshold

def johansen_cointegration_test(ts, window=50, significance_level=0.05):
    n_obs, n_variables = ts.shape
    cointegration_results = []
    
    for i in range(window, n_obs):
        window_data = ts[i-window:i]

        # Check for constant or low-variance columns
        if np.any([is_constant(window_data[:, var]) or is_low_variance(window_data[:, var]) for var in range(n_variables)]):
            cointegration_results.append(False)  # Skip test if any column is constant or near-constant
            continue

        try:
            johansen_test = coint_johansen(window_data, det_order=0, k_ar_diff=1)
            trace_statistic = johansen_test.lr1
            critical_values = johansen_test.cvt[:, 1]  # 5% significance level
            
            cointegration_results.append(np.any(trace_statistic > critical_values))
        except np.linalg.LinAlgError:
            cointegration_results.append(False)  # Skip test if matrix inversion fails
    
    return np.array(cointegration_results)

def rolling_tests_multivariate(ts, window=50, alpha=0.05):
    n_variables = ts.shape[1]
    adf_results, kpss_results, pp_results, sigma_results = [], [], [], []
    
    for i in range(window, len(ts)):
        adf_p_values, kpss_p_values, pp_p_values, sigma_flags = [], [], [], []
        
        for var in range(n_variables):
            window_data = ts[i-window:i, var]
            
            if is_constant(window_data):
                adf_p_values.append(1)  # ADF p-value = 1 (not stationary)
                kpss_p_values.append(0)  # KPSS p-value = 0 (stationary)
                pp_p_values.append(1)  # PP p-value = 1 (not stationary)
                sigma_flags.append(False)  # No variance increase
                continue
            
            adf_p = adfuller(window_data)[1]
            kpss_p = kpss(window_data, regression='c')[1]
            
            try:
                pp_p = PhillipsPerron(window_data).pvalue  # Run PP test safely
            except:
                pp_p = 1  # Treat constant or singular cases as non-stationary
            
            adf_p_values.append(adf_p)
            kpss_p_values.append(kpss_p)
            pp_p_values.append(pp_p)
            
            # 2-Sigma Method (Dakos et al.)
            window_std = np.std(window_data)
            mean_std = np.mean([np.std(ts[max(0, j-window):j, var]) for j in range(i-window, i)])
            sigma_flags.append(window_std > 2 * mean_std)
        
        adf_results.append(np.array(adf_p_values) < alpha)  # True if stationary
        kpss_results.append(np.array(kpss_p_values) > alpha)  # True if stationary
        pp_results.append(np.array(pp_p_values) < alpha)  # True if stationary
        sigma_results.append(np.array(sigma_flags))  # Flag if variance > 2-sigma threshold
    
    return np.array(adf_results), np.array(kpss_results), np.array(pp_results), np.array(sigma_results)
```

```python
def plot_results_multivariate(ts, evectors_100, evectors_60, change_points, window, title, figtitle):
    # Perform rolling Johansen cointegration test
    cointegration_results = johansen_cointegration_test(evectors_100, window)

    # Perform rolling stationarity tests
    adf_results, kpss_results, pp_results, sigma_results = rolling_tests_multivariate(evectors_100, window)

    n_variables = evectors_100.shape[1]
    fig, axes = plt.subplots(5, 2, figsize=(24, 20), sharex=True)
    
    tests = [("ADF", adf_results), ("KPSS", kpss_results), ("PP", pp_results), ("2-Sigma", sigma_results), ("Cointegration", cointegration_results)]
    colors = ['red', 'blue', 'green', 'purple', 'orange']
    
    ts['rolling'] = ts['value'].rolling(7).mean()
    ts = ts.reset_index()
    # Plot each test result for each variable
    for ax, (test_name, results), color in zip(axes[:,0], tests, colors):
        ax.plot(ts['date'],ts['value'],'.',color='#2A1E5C')
        ax.plot(counts['date'],counts['rolling'],color='#D62828')
        # Mark non-stationary periods for each variable
        if test_name != "Cointegration":
            for var in range(n_variables):
                ax.scatter(ts['date'][window:][results[:, var]], np.ones(np.sum(results[:, var])) * var, 
                           color=color, alpha=0.5, label=f'{test_name} Non-Stationary Var {var+1}' if var == 0 else "")
        else:
            # For Cointegration, plot where cointegration is True
            ax.scatter(ts['date'][window:][cointegration_results], np.zeros(np.sum(cointegration_results)), 
                       color=color, alpha=0.7, label=f'{test_name} Cointegrated')

        # Mark change points
        for cp in change_points:
            ax.axvline(cp, color='black', linestyle='--', alpha=0.7, label='Change Point' if cp == change_points[0] else "")
        
        ax.set_title(f"100 days - {title}: {test_name}")
        ax.legend()

    # Perform rolling Johansen cointegration test
    cointegration_results = johansen_cointegration_test(evectors_60, window)

    # Perform rolling stationarity tests
    adf_results, kpss_results, pp_results, sigma_results = rolling_tests_multivariate(evectors_60, window)

    tests = [("ADF", adf_results), ("KPSS", kpss_results), ("PP", pp_results), ("2-Sigma", sigma_results), ("Cointegration", cointegration_results)]
    
    # Plot each test result for each variable
    for ax, (test_name, results), color in zip(axes[:,1], tests, colors):
        ax.plot(ts['date'],ts['value'],'.',color='#2A1E5C')
        ax.plot(counts['date'],counts['rolling'],color='#D62828')
        # Mark non-stationary periods for each variable
        if test_name != "Cointegration":
            for var in range(n_variables):
                ax.scatter(ts['date'][window:][results[:, var]], np.ones(np.sum(results[:, var])) * var, 
                           color=color, alpha=0.5, label=f'{test_name} Non-Stationary Var {var+1}' if var == 0 else "")
        else:
            # For Cointegration, plot where cointegration is True
            ax.scatter(ts['date'][window:][cointegration_results], np.zeros(np.sum(cointegration_results)), 
                       color=color, alpha=0.7, label=f'{test_name} Cointegrated')

        # Mark change points
        for cp in change_points:
            ax.axvline(cp, color='black', linestyle='--', alpha=0.7, label='Change Point' if cp == change_points[0] else "")
        
        ax.set_title(f"60 days - {title}: {test_name}")
        ax.legend()

    plt.xlabel("Time")
    plt.savefig(f"Figures/{figtitle}.pdf",dpi=2400,bbox_inches='tight')
    plt.show()
```

```python
def is_constant(series):
    return np.all(series == series[0])

def is_low_variance(series, threshold=1e-5):
    return np.var(series) < threshold

def johansen_cointegration_test(ts, window=50, significance_level=0.05):
    n_obs, n_variables = ts.shape
    cointegration_results = []
    
    for i in range(window, n_obs):
        window_data = ts[i-window:i]

        # Check for constant or low-variance columns
        if np.any([is_constant(window_data[:, var]) or is_low_variance(window_data[:, var]) for var in range(n_variables)]):
            cointegration_results.append(False)  # Skip test if any column is constant or near-constant
            continue

        try:
            johansen_test = coint_johansen(window_data, det_order=0, k_ar_diff=1)
            trace_statistic = johansen_test.lr1
            critical_values = johansen_test.cvt[:, 1]  # 5% significance level
            
            cointegration_results.append(np.any(trace_statistic > critical_values))
        except np.linalg.LinAlgError:
            cointegration_results.append(False)  # Skip test if matrix inversion fails
    
    return np.array(cointegration_results)

def rolling_tests_multivariate(ts, window=50, alpha=0.05):
    n_variables = ts.shape[1]
    adf_results, kpss_results, pp_results, sigma_results = [], [], [], []
    
    for i in range(window, len(ts)):
        adf_p_values, kpss_p_values, pp_p_values, sigma_flags = [], [], [], []
        
        for var in range(n_variables):
            window_data = ts[i-window:i, var]
            
            if is_constant(window_data):
                adf_p_values.append(1)  # ADF p-value = 1 (not stationary)
                kpss_p_values.append(0)  # KPSS p-value = 0 (stationary)
                pp_p_values.append(1)  # PP p-value = 1 (not stationary)
                sigma_flags.append(False)  # No variance increase
                continue
            
            adf_p = adfuller(window_data)[1]
            kpss_p = kpss(window_data, regression='c')[1]
            
            try:
                pp_p = PhillipsPerron(window_data).pvalue  # Run PP test safely
            except:
                pp_p = 1  # Treat constant or singular cases as non-stationary
            
            adf_p_values.append(adf_p)
            kpss_p_values.append(kpss_p)
            pp_p_values.append(pp_p)
            
            # 2-Sigma Method (Dakos et al.)
            window_std = np.std(window_data)
            mean_std = np.mean([np.std(ts[max(0, j-window):j, var]) for j in range(i-window, i)])
            sigma_flags.append(window_std > 2 * mean_std)
        
        adf_results.append(np.array(adf_p_values) < alpha)  # True if stationary
        kpss_results.append(np.array(kpss_p_values) > alpha)  # True if stationary
        pp_results.append(np.array(pp_p_values) < alpha)  # True if stationary
        sigma_results.append(np.array(sigma_flags))  # Flag if variance > 2-sigma threshold
    
    return np.array(adf_results), np.array(kpss_results), np.array(pp_results), np.array(sigma_results)

def plot_results_multivariate(ts, evectors_100, evectors_60, change_points, window, title, figtitle):
    # Perform rolling Johansen cointegration test
    cointegration_results = johansen_cointegration_test(evectors_100, window)

    # Perform rolling stationarity tests
    adf_results, kpss_results, pp_results, sigma_results = rolling_tests_multivariate(evectors_100, window)

    n_variables = evectors_100.shape[1]
    fig, axes = plt.subplots(5, 2, figsize=(24, 20), sharex=True)
    
    tests = [("ADF", adf_results), ("KPSS", kpss_results), ("PP", pp_results), ("2-Sigma", sigma_results), ("Cointegration", cointegration_results)]
    colors = ['red', 'blue', 'green', 'purple', 'orange']
    
    ts['rolling'] = ts['value'].rolling(7).mean()
    ts = ts.reset_index()
    # Plot each test result for each variable
    for ax, (test_name, results), color in zip(axes[:,0], tests, colors):
        ax.plot(ts['date'],ts['value'],'.',color='#2A1E5C')
        ax.plot(counts_nhs['date'],counts_nhs['rolling'],color='#D62828')
        # Mark non-stationary periods for each variable
        if test_name != "Cointegration":
            for var in range(n_variables):
                ax.scatter(ts['date'][window:][results[:, var]], np.ones(np.sum(results[:, var])) * var, 
                           color=color, alpha=0.5, label=f'{test_name} Non-Stationary Var {var+1}' if var == 0 else "")
        else:
            # For Cointegration, plot where cointegration is True
            ax.scatter(ts['date'][window:][cointegration_results], np.zeros(np.sum(cointegration_results)), 
                       color=color, alpha=0.7, label=f'{test_name} Cointegrated')

        # Mark change points
        for cp in change_points:
            ax.axvline(cp, color='black', linestyle='--', alpha=0.7, label='Change Point' if cp == change_points[0] else "")
        
        ax.set_title(f"100 days - {title}: {test_name}")
        ax.legend()

    # Perform rolling Johansen cointegration test
    cointegration_results = johansen_cointegration_test(evectors_60, window)

    # Perform rolling stationarity tests
    adf_results, kpss_results, pp_results, sigma_results = rolling_tests_multivariate(evectors_60, window)

    tests = [("ADF", adf_results), ("KPSS", kpss_results), ("PP", pp_results), ("2-Sigma", sigma_results), ("Cointegration", cointegration_results)]
    
    # Plot each test result for each variable
    for ax, (test_name, results), color in zip(axes[:,1], tests, colors):
        ax.plot(ts['date'],ts['value'],'.',color='#2A1E5C')
        ax.plot(counts_nhs['date'],counts_nhs['rolling'],color='#D62828')
        # Mark non-stationary periods for each variable
        if test_name != "Cointegration":
            for var in range(n_variables):
                ax.scatter(ts['date'][window:][results[:, var]], np.ones(np.sum(results[:, var])) * var, 
                           color=color, alpha=0.5, label=f'{test_name} Non-Stationary Var {var+1}' if var == 0 else "")
        else:
            # For Cointegration, plot where cointegration is True
            ax.scatter(ts['date'][window:][cointegration_results], np.zeros(np.sum(cointegration_results)), 
                       color=color, alpha=0.7, label=f'{test_name} Cointegrated')

        # Mark change points
        for cp in change_points:
            ax.axvline(cp, color='black', linestyle='--', alpha=0.7, label='Change Point' if cp == change_points[0] else "")
        
        ax.set_title(f"60 days - {title}: {test_name}")
        ax.legend()

    plt.xlabel("Time")
    plt.savefig(f"Figures/{figtitle}.pdf",dpi=2400,bbox_inches='tight')
    plt.show()
```
